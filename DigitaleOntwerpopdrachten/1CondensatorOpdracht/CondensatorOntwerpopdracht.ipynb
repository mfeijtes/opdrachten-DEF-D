##### Antwoord 3b

Vanuit Gauss's Law weten we:
\
$\vec{\nabla} \cdot \vec{E} = \frac{\rho}{\epsilon _0} $, en voor een twee dimensionale schijf:
\
$\vec{\nabla} \cdot \vec{E} = \frac{\sigma}{\epsilon _0} \Rightarrow \sigma = \epsilon _0(\vec{\nabla} \cdot \vec{E})$
\
Dit uitgewerkt geeft:
\
$\sigma =\epsilon _0 (
\frac{1}{r}\frac{\partial}{\partial r}\left(r E_r\right)
+
\frac{1}{r}\frac{\partial E_\theta}{\partial \theta}
+
\frac{\partial E_z}{\partial z})=\epsilon _0(\frac{E_r}{r} + \frac{\partial E_r}{\partial r})$
\
Dit wijst op een singulariteit bij $r=0$, neem daarom de taylor expansie rond $r=0$:
\
$E_r(r) \approx ar$
\
Hieruit volgt:
\
$\vec{\nabla} \cdot \vec{E} = \frac{1}{r}\frac{\partial (rE_r)}{\partial r} \approx \frac{1}{r}\frac{\partial (ar^2)}{\partial r} = 2a$
\
Maar $a = \frac{\partial E_r}{\partial r} \Bigr\rvert_{r=0}$, dus:
\
\
$\sigma (0) \approx \epsilon _0 2 \frac{\partial E_r}{\partial r} \Bigr\rvert_{r=0} \approx \epsilon _0 2 \frac{\Delta E_r}{\Delta r} = \epsilon _0 2 \frac{E_r(r_1)}{r_1}$
\
Hier wordt $r_1$ erg dicht bij 0 gekozen.



````python
# Neem eerste roosterpunt weg van r = 0
r1 = r[idx_R + 1]
Er1 = Er[idx_z_zero, idx_R + 1]   # E_r component

# Divergentie bij r = 0
divE_0 = 2 * Er1 / r1

# Oppervlaklading
sigma0 = epsilon0 * divE_0

print("Oppervlaktelading in het midden van de schijf:", sigma0, "C/m^2")
````

#### Vraag 3c: De oppervlaktelading van een geleidende schijf op constant potentiaal

De oppervlakte lading $\sigma(r)$ van een geleidende schijf met straal $R$ is niet uniform verdeeld over de schijf, maar wordt gegeven door (Problems 2.58 and 2.63 Griffiths 5th edition):
$$\sigma(r) = \frac{Q}{2 \pi R \sqrt{R^2 - r^2}}$$

1. Bereken de totale lading op de schijf met behulp van vraag 3a en 3b.

4. Plot in hetzelfde figuur de oppervlaktelading van een uniform geladen schijf met dezelfde lading $Q$.


````python
#Herschrijf de formule om op te lossen voor Q en zet r=0
Q = 2*np.pi*disk_radius**2*sigma0
print("Totale lading op de schijf:", Q, "C")
#Voor een schijf met uniforme oppervlaktelading is de oppervlaktelading gelijk aan de totale lading gedeeld door het oppervlak van de schijf, dus:
sigma_uniform = Q / (np.pi * disk_radius**2)
print("Oppervlaktelading voor een uniforme ladingverdeling:", sigma_uniform, "C/m^2")
#Deze is exact aan het dubbele van een geleidende schijf
def sigma_geleidende_schijf(r, disk_radius):
    sigma = Q/(2*np.pi*disk_radius*np.sqrt(disk_radius**2 - r**2))
    return sigma

r_values = np.linspace(0, disk_radius-0.01, 100)
sigma_values = sigma_geleidende_schijf(r_values, disk_radius)
sigma_uniform_values = np.full_like(r_values, sigma_uniform)

plt.figure()
plt.plot(r_values, sigma_values, label="Oppervlakteladingverdeling over de schijf")
plt.plot(r_values,sigma_uniform_values, label="Oppervlakteladingverdeling voor een uniforme ladingverdeling", linestyle="--")
plt.xlabel("r (m)")
plt.ylabel("Oppervlaktelading (C/m^2)")
plt.title("Oppervlakteladingverdeling op een geleidende schijf")
plt.legend()
plt.show()
````
